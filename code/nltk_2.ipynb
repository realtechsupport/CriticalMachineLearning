```python
# Webtext parsing functions
def tag_visible(element):
    if element.parent.name in ['style', 'script', 'head', 'title', 'meta', '[document]']:
        return False
    if isinstance(element, Comment):
        return False
    return True

def text_from_html(body):
    soup = BeautifulSoup(body, 'html.parser')
    texts = soup.findAll(string=True)
    visible_texts = filter(tag_visible, texts)
    return (u" ".join(t.strip() for t in visible_texts))
```

```python
# Get printable text from the url
# Some sites will be blocked (403 error)
# Non-english information can also be collected

#English
#url = "https://www.nytimes.com/"
#German
#url = "https://www.faz.net/"
#Arabic
#url = "https://arabic.cgtn.com/home"
#Korean
url = "https://m.10000recipe.com/"

#Read the content of the url into a variable
html = urllib.request.urlopen(url).read()
#Get the printable text from the variable containing the complete content
webtext = text_from_html(html)

print(webtext)
```

```python
#Find the most frequent words
from collections import Counter

split = webtext.split()
counter = Counter(split)
most_frequent = counter.most_common(5)
print(most_frequent)
```

Lets write some code to get unique items in a list of text items. We can imagine that the input text items are the result of web scrapping, produced with a utility like Beautiful Soup.


```python
nwords = ['house', 'house', 'car', 'mouse', 'cat', 'dog', 'home', 'home@']

#Here we are using the "in" operation to compare each item in the list of nwords with those in the list unique

unique = []
for item in nwords:
  if (item in unique):
    pass
  else:
    unique.append(item)

print(nwords)
print(unique)

# This approach does not know that home and home@ are infact the same items
```

```python
nwords = ['house', 'house', 'car', 'mouse', 'cat', 'dog', 'home', 'home@']

#Let use regular expressions to filter out the symbol @
#Regular expressions are a sequence of characters that specifies a match pattern in text.

#Regular expressions in python: https://docs.python.org/3/library/re.html

#import regular expressions
import re

# [^\w] will match anything that's not alphanumeric or underscore
expression = r'[^\w]'
# replace with nothing
replacement = ''

unique = []
for item in nwords:
  clean_item = re.sub(expression, replacement, item)
  if (clean_item in unique):
    pass
  else:
    unique.append(clean_item)

print(unique)
```

```python
import re

input_string = "Please join this meeting using this code: %#7878 at 09:15 PST"

# Define the regular expression pattern
timestamp_regex = r"\d{2}:\d{2} \w{3}"
#pattern = r":(.*?)at"
#pattern = r"#%\d{4}"
#pattern = r"#%(.*?)"
#pattern = r"^%#(.*?)\Z"
pattern = r":\s*([^a-zA-Z\s]+)\s*at"
```

```python
# Match the timestamp using the regular expression
timestamp_match = re.search(timestamp_regex, input_string)

# Extract the timestamp if it was found
if timestamp_match:
    timestamp = timestamp_match.group()
    print("Timestamp:", timestamp)
else:
    print("Timestamp not found")
```

```python
import re

def extract_meeting_code(input_string):
    # Define a regular expression pattern to match the meeting code
    #pattern = r"code:\s*%#(\d+)"
    #pattern =r"code:\s*%#(\S+)"
    #pattern = r"code:\s*%#([^\s]+)"
    pattern = r":\s*([^a-zA-Z\s]+)\s*at"

    # Use re.search to find the match in the input string
    match = re.search(pattern, input_string)

    # Check if a match is found
    if match:
        # Extract and return the meeting code
        meeting_code = match.group(1)
        return meeting_code
    else:
        # Return None if no match is found
        return None

# Example usage
input_string = "Please join this meeting using this code: %#7878 at 09:15 PST"
meeting_code = extract_meeting_code(input_string)

if meeting_code:
    print("Meeting Code:", meeting_code)
else:
    print("No meeting code found.")
```
